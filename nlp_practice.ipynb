```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
import pandas as pd
```

```python
data=pd.read_csv('/content/news.csv')
data.head()
```

```python
dataset=data.iloc[:, 0]
dataset
```

```python
# sentence=[]
# for i in range(len(dataset)):
#   sentence+=nltk.sent_tokenize(dataset.iloc[i])
sentence=nltk.sent_tokenize(dataset.iloc[0])
sentence
```

```python
# lamitizing
lamitizer=WordNetLemmatizer()
for i in range(len(sentence)):
  words=nltk.word_tokenize(sentence[i])
  words=[lamitizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
  sentence[i]=' '.join(words)
sentence
```
